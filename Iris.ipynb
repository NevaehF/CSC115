```python
### Import the pandas library 

import pandas as pd # Pandas is iused for workingwith tabular data (ros columns) 
import numpy as np # Numpy is used for numerical computations, arrays, and math functions) 
import matplotlib.pyplot # Matplot used for plotting our charts 

from sklearn.datasets import load_iris
from scipy import stats # extra statistical functions (mode)
```

```python
# Load the Iris Dataset 

iris = load_iris() # Dictionary like object with data, targets, and feature names
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
```

```python
iris_df.head()
```

```python
iris_df.info()
```

```python
# 1) Calculate the mean, median, and mode 

iris_df.mean() # average values of each column (feature)
```

```python
# 2) Median

iris_df.median()
```

```python
# 3) Mode

iris_df.mode().loc[0] # Take the first row with each loc because mode () can return multiple
```

```python
# 4) Range = The difference between the max and min values 

iris_df.max() - iris_df.min()
```

```python
# 5) Variance - measure of spread ( average squared deviation from the mean) 

iris_df.var()
```

```python
# 6 ) Standard deviation - squared root of the variance, also measures the spread

iris_df.std()
```

```python
# 7 ) Percentiles (25th, 50th, 75th) 

iris_df.quantile([0.25, 0.5, 0.75])
```
