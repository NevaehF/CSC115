```python

# If needed, install dependencies (uncomment in your local environment)
# !pip install requests beautifulsoup4 pandas lxml

import time
import requests
from bs4 import BeautifulSoup
import pandas as pd
from pathlib import Path

BASE_URL = "https://quotes.toscrape.com/"
headers = {
    "User-Agent": "Mozilla/5.0 (X11; Linux x86_64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}
```

```python
def scrape_quotes(base_url=BASE_URL, headers=headers, pause_s=1.0, max_pages=50):
    """Scrape quotes.toscrape.com with polite pacing and pagination.
    Returns a list of dicts.
    """
    records = []
    url = base_url
    pages = 0

    while url and pages < max_pages:
        resp = requests.get(url, headers=headers, timeout=15)
        resp.raise_for_status()  # raise an error if status != 200
        soup = BeautifulSoup(resp.text, "lxml")

        # Each quote is within <div class="quote">
        for q in soup.select("div.quote"):
            text = q.select_one("span.text")
            author = q.select_one("small.author")
            tag_els = q.select("div.tags a.tag")
            tags = [t.get_text(strip=True) for t in tag_els] if tag_els else []

            records.append({
                "quote": text.get_text(strip=True) if text else None,
                "author": author.get_text(strip=True) if author else None,
                "tags": tags
            })

        # Find "Next" page link
        next_link = soup.select_one("li.next a")
        if next_link and next_link.get("href"):
            url = base_url.rstrip("/") + next_link.get("href")
        else:
            url = None

        pages += 1
        time.sleep(pause_s)  # polite pause to avoid hammering the server

    return records

# Run the scraper (uncomment below when running locally with internet)
data = scrape_quotes()
df = pd.DataFrame(data)
df.head()
```
